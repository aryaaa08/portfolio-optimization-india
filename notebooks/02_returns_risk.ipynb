```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
prices = pd.read_csv(
    "adjusted_prices.csv",
    index_col=0,
    parse_dates=True
)
```

```python
prices.head()
```

```python
prices.plot(
    figsize=(14, 8),
    title="Historical Adjusted Prices of Selected Indian Equities"
)

plt.xlabel("Date")
plt.ylabel("Adjusted Price")
plt.show()
```

```python
normalized_prices = prices / prices.iloc[0] * 100
```

```python
normalized_prices.head()
```

```python
normalized_prices.plot(
    figsize=(14, 8),
    title="Normalized Price Performance (Base = 100)"
)

plt.xlabel("Date")
plt.ylabel("Normalized Price")
plt.show()
```

```python
daily_returns = prices.pct_change()
daily_returns.head()
```

```python
daily_returns = daily_returns.dropna()
daily_returns.head()
```

```python
daily_returns.describe()
```

```python
daily_returns.plot(
    figsize=(14, 8),
    title="Daily Returns - Indian Equities",
    alpha=0.7
)

plt.xlabel("Date")
plt.ylabel("Daily Return")
plt.show()
```

## Calculate average daily return

```python
mean_daily_return = daily_returns.mean()
mean_daily_return
```

## Annualize the return

```python
annualized_return = (1 + mean_daily_return) ** 252 - 1
annualized_return_pct = annualized_return * 100
annualized_return_pct.sort_values(ascending=False)
```

## Variance

```python
daily_variance = daily_returns.var()
daily_variance
```

```python
daily_volatility = daily_returns.std()
```

```python
annualized_volatility = daily_volatility * np.sqrt(252)
annualized_volatility_pct = annualized_volatility * 100
annualized_volatility_pct.sort_values(ascending=False)
```

```python
risk_return = pd.DataFrame({
    "Annualized Return (%)": annualized_return_pct,
    "Annualized Volatility (%)": annualized_volatility_pct
})
```

```python
risk_return.sort_values(
    "Annualized Return (%)",
    ascending=False
)
```

```python
plt.figure(figsize=(12, 8))

plt.scatter(
    annualized_volatility * 100,
    annualized_return * 100
)

for stock in risk_return.index:
    plt.annotate(
        stock.replace(".NS", ""),
        (
            annualized_volatility[stock] * 100,
            annualized_return[stock] * 100
        )
    )

plt.xlabel("Annualized Volatility (%)")
plt.ylabel("Annualized Return (%)")
plt.title("Risk–Return Profile of Selected Indian Equities")

plt.show()
```

```python
covariance_matrix = daily_returns.cov()
covariance_matrix
```

```python
annualized_covariance = covariance_matrix * 252
```

```python
annualized_covariance.to_csv(
    "covariance_matrix.csv"
)
```

```python
correlation_matrix = daily_returns.corr()
correlation_matrix
```

```python
plt.figure(figsize=(14, 11))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Selected Indian Equities")
plt.show()
```
